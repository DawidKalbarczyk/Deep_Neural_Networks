```python
import importlib

biblioteki = ["numpy", "pandas", "matplotlib", "sklearn", "skimage", "cv2", "PIL",
              "rasterio", "albumentations", "torch", "torchvision", "wandb", "ultralytics"]

for nazwa in biblioteki:
    try:
        modul = importlib.import_module(nazwa)
        print(f"OK    {nazwa:15s} {getattr(modul, '__version__', '')}")
    except Exception as e:
        print(f"BRAK  {nazwa:15s} -> pip install (sprawdź nazwę pakietu w environment.yml)")
```

```python
import torch

print("CUDA dostępna:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("Karta:", torch.cuda.get_device_name(0))

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
x = torch.rand(1000, 1000, device=device)
print("Test obliczeń na", device, "->", float((x @ x).mean()))
```
